```python
# @title 2. Instalación de Mambaforge y Nextflow (Forzada)
# Esta celda garantiza que el PATH se actualice correctamente.

# 1. Instalar Mambaforge
print("--- Instalando Mambaforge ---")
!wget -qO- https://github.com/conda-forge/miniforge/releases/latest/download/Mambaforge-Linux-x86_64.sh > miniforge.sh
!bash miniforge.sh -b -p /opt/conda

# 2. **ACTIVACIÓN FORZADA DEL PATH:** Añadir el directorio bin de Conda al PATH
import os
os.environ['PATH'] = '/opt/conda/bin:' + os.environ['PATH']
print(f"Nuevo PATH: {os.environ['PATH']}")

# 3. Instalar Nextflow
print("\n--- Instalando Nextflow ---")
!conda install -c bioconda nextflow -y

# 4. Verificar la instalación
print("\n--- Versión de Nextflow ---")
!nextflow -version

print("\n🎉 Instalación completa. El entorno está listo.")
```

```python
# @title 2.5. Creación del Archivo de Acceso SRA (Códigos de Ejemplo)

# Códigos SRA de ejemplo que representan las muestras que Nextflow descargará
sra_codes = """
SRR860001
SRR860002
"""

# Escribir la lista en el archivo requerido por el workflow
# La ruta es 'DATABASE/SRA_accession_list.txt'
with open("DATABASE/SRA_accession_list.txt", "w") as f:
    f.write(sra_codes.strip())

print("Archivo DATABASE/SRA_accession_list.txt creado con éxito.")
print("--- Contenido del Archivo ---")
# Mostrar el contenido del archivo recién creado para verificar
!cat DATABASE/SRA_accession_list.txt
```

```python
# @title 3. Ejecución del Workflow de Filogenómica Mitocondrial (INTENTO FINAL)

print("Iniciando la ejecución del workflow sin perfil Conda estricto. Esto debe iniciar la descarga de datos SRA.")

!nextflow run scripts/main_workflow.nf \
    --max-cpus 4 \
    --data_list 'DATABASE/SRA_accession_list.txt' \
    -w /content/work \
    -N PROYECTO-MACACO.report
```
